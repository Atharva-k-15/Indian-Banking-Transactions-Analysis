```python
print("Duplicate rows:", df.duplicated().sum())
print("Missing values:\n", df.isnull().sum())
print("Duplicate Transaction IDs:", df["transaction_id"].duplicated().sum())
```

```python
# 1. Check date parsing
df["transaction_date"] = pd.to_datetime(
    df["transaction_date"], format="%d/%m/%Y", errors="coerce"
)
print("Invalid dates:", df["transaction_date"].isna().sum())
```

```python
# 2. Check valid ranges
print("Invalid transaction hours:",
      (~df["transaction_hour"].between(0, 23)).sum())

print("Invalid fraud flags:",
      (~df["is_fraud"].isin([0, 1])).sum())
```

```python
# 3. Check loan-related consistency
print(pd.crosstab(df["has_loan"], df["loan_type"], dropna=False))
```

```python
# 4. Inspect the largest transaction amounts
print(df.nlargest(10, "transaction_amount")[
    ["transaction_id", "transaction_amount",
     "transaction_type", "transaction_status"]
])
```

```python
# Records where a loan exists but loan_type is missing
problem_records = df[
    (df["has_loan"] == 1) &
    (df["loan_type"].isna())
]

print("Problem records:", len(problem_records))
print("Percentage of dataset:",
      round(len(problem_records) / len(df) * 100, 2), "%")
```

```python
plt.figure(figsize=(12, 6))

order = df["transaction_type"].value_counts().index

sns.countplot(
    data=df,
    x="transaction_type",
    order=order
)

plt.title("Distribution of Banking Transaction Types")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
status_counts = df["transaction_status"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    status_counts,
    labels=status_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Transaction Status Distribution")
plt.axis("equal")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="transaction_amount",
    bins=50,
    kde=True
)

plt.title("Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 7))

sns.boxplot(
    data=df,
    x="transaction_type",
    y="transaction_amount"
)

plt.title("Transaction Amounts by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Transaction Amount")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Ensure transaction_date is a datetime column
df["transaction_date"] = pd.to_datetime(
    df["transaction_date"],
    errors="coerce",
    dayfirst=True
)

monthly_transactions = (
    df.dropna(subset=["transaction_date"])
      .set_index("transaction_date")
      .resample("ME")
      .size()
)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_transactions.index,
    monthly_transactions.values,
    marker="o",
    linewidth=2
)

plt.title("Monthly Banking Transaction Volume")
plt.xlabel("Month")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
numeric_columns = [
    "transaction_amount",
    "account_balance",
    "credit_score",
    "emi_amount",
    "transaction_hour",
    "is_fraud"
]

correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Banking Variables")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="channel",
    hue="is_fraud"
)

plt.title("Fraud Flags by Transaction Channel")
plt.xlabel("Transaction Channel")
plt.ylabel("Number of Transactions")
plt.legend(title="Fraud Flag", labels=["Not Flagged", "Flagged"])
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

```python
fraud_rate = (
    df.groupby("channel")["is_fraud"]
      .mean()
      .mul(100)
      .sort_values(ascending=False)
      .reset_index(name="fraud_rate_pct")
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=fraud_rate,
    x="channel",
    y="fraud_rate_pct"
)

plt.title("Percentage of Transactions Flagged for Fraud by Channel")
plt.xlabel("Transaction Channel")
plt.ylabel("Flagged Transaction Rate (%)")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

```python
status_by_channel = pd.crosstab(
    df["channel"],
    df["transaction_status"]
)

status_by_channel.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 6)
)

plt.title("Transaction Status by Channel")
plt.xlabel("Channel")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=30, ha="right")
plt.legend(title="Transaction Status", bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()
```
